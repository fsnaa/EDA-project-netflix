```python
import pandas as pd
df=pd.read_csv(r"C:\Users\HP\Downloads\netflix_titles.csv")
df
```

```python
df.duplicated().sum()
df.drop_duplicates(inplace=True)
```

```python
df['date_added'] = df['date_added'].str.strip()
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
```

```python
df.isnull().sum()
```

```python
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Not Available')

```

```python
df['country'] = df['country'].fillna(df['country'].mode()[0])
df['rating'] = df['rating'].fillna(df['rating'].mode()[0])
```

```python
df.dropna(subset=['date_added'], inplace=True)
```

```python
df['type'] = df['type'].str.strip()
df['country'] = df['country'].str.strip()
df['rating'] = df['rating'].str.strip()
```

```python
df['duration_num'] = df['duration'].str.extract('(\d+)').astype(float)
```

```python
df = df[~df['rating'].str.contains('min', na=False)]
df
```

```python
Q1 = df['release_year'].quantile(0.25)
Q3 = df['release_year'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['release_year'] < lower) | (df['release_year'] > upper)]
outliers
```

## Data Cleaning Summary

- Removed duplicate records  
- Converted date_added to datetime format  

- Handled missing values:  
  - director → filled with 'Unknown'  
  - cast → filled with 'Not Available'  
  - country → filled with mode  
  - rating → filled with mode  
  - date_added → removed missing rows  

- Standardized categorical columns  

- Extracted numerical values from duration column  



## Outlier Detection


Outliers were detected using the IQR method on the release_year column. Although a number of values were identified as outliers, they represent valid older content and were not removed, as they are important for understanding the full distribution of release years.
